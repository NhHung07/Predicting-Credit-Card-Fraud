# XGBoost v2 & v3 — GLM-Boosted & Standalone Digit-TE GPU Trees

- **Thành viên:** B (Cao Anh Minh)
- **Họ thuật toán:** Classical Machine Learning (XGBoost `hist` GPU)
- **Kỹ thuật nâng cấp:** Kết hợp bộ đặc trưng `Digit-TE` + `base_margin` từ logit GLM, giới hạn độ sâu `max_depth=3` chống overfit.
  - `xgb_b_v2` (GLM-Boosted `max_depth=3`, `lr=0.02`): CV **0.95073**, Public LB **0.95110** (Best Single Model LB!)
  - `xgb_b_v3` (Standalone `max_depth=3`, `lr=0.03`): CV **0.94996**

> Mã nguồn thực thi đầy đủ: `B/train_xgb_v2.py & B/train_xgb_v3.py`

```
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score

y_true = pd.read_csv("../data/train.csv", usecols=["is_fraud"])["is_fraud"].values
for exp_id in ["xgb_b_v2", "xgb_b_v3"]:
    oof = np.load(f"outputs/{exp_id}_oof.npy")
    print(f"{exp_id} Overall OOF ROC-AUC: {roc_auc_score(y_true, oof):.5f}")
```
