# PyTorch Wide & Deep Periodic TabResNet (Deep Learning thuần túy v2)

- **Thành viên:** C (Phạm Hồng Nam)
- **Họ thuật toán:** Deep Learning (PyTorch GPU, 100% không dùng LLM)
- **Kiến trúc nâng cấp:**
  1. `QuantileTransformer(normal)` + `PeriodicEmbedding` (sin/cos đa tần số học được theo Gorishniy et al., NeurIPS 2022) cho biến số liên tục.
  2. `nn.Embedding` cho biến định tính.
  3. Kết nối `Wide` tuyến tính skip + Khối `ResNet` với `LayerNorm`, `SiLU`, `Dropout(0.45)` và nhiễu Gaussian đầu vào (`0.05`) chống overfit.
- **Kết quả:** Mean Fold CV **0.95021** (tăng từ 0.94180), Public LB **0.95052** (+0.00726 so với `mlp_c_v1` 0.94326!).

> Mã nguồn thực thi đầy đủ: `C/train_tabresnet_v2.py`

```
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score

y_true = pd.read_csv("../data/train.csv", usecols=["is_fraud"])["is_fraud"].values
folds = pd.read_csv("../data/folds.csv")["fold"].values
oof = np.load("outputs/mlp_c_v2_oof.npy")
f_aucs = [roc_auc_score(y_true[folds==f], oof[folds==f]) for f in range(5)]
print(f"Mean Fold AUC: {np.mean(f_aucs):.5f} +/- {np.std(f_aucs):.5f}")
```
