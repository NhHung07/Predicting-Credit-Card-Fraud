# LightGBM v3 & v4 — Leak-Free Digit-TE & GLM-Boosted Shallow Trees

- **Thành viên:** A (Trần Nhật Hưng)
- **Họ thuật toán:** Classical Machine Learning (GBDT)
- **Kỹ thuật nâng cấp:** Sử dụng `shared/fe_digit_te.py` trích xuất đặc trưng chữ số (cents, tenths, int, div10/100/1000, first1/2/3), Frequency Encoding và Target Encoding chống rò rỉ (inner 5-fold m-estimate).
  - `lgbm_a_v3` (GLM-Boosted `num_leaves=4`, `lr=0.02`): CV **0.95084**, Public LB **0.95094**
  - `lgbm_a_v4` (Standalone `num_leaves=6`, `lr=0.025`): CV **0.95022**, Public LB **0.95089**

> Mã nguồn thực thi đầy đủ: `A/train_lgbm_v3.py & A/train_lgbm_v4.py`

```
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score

y_true = pd.read_csv("../data/train.csv", usecols=["is_fraud"])["is_fraud"].values
for exp_id in ["lgbm_a_v3", "lgbm_a_v4"]:
    oof = np.load(f"outputs/{exp_id}_oof.npy")
    print(f"{exp_id} Overall OOF ROC-AUC: {roc_auc_score(y_true, oof):.5f}")
```
