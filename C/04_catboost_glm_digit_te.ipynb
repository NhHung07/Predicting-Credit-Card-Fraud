# CatBoost v2 & Logistic GLM — Digit-TE Pure ML Models

- **Thành viên:** C (Phạm Hồng Nam)
- **Họ thuật toán:** Classical Machine Learning (Linear GLM & CatBoost Oblivious Trees)
- **Kết quả:**
  - `glm_c_v1` (Logistic Regression L2 trên TE Logits + QuantileTransformer): CV **0.94981**, Public LB **0.94984**
  - `cat_c_v2` (GLM-Boosted CatBoost `depth=3`, `lr=0.04`): CV **0.95062**, Public LB **0.95064**

> Mã nguồn thực thi đầy đủ: `C/train_glm_v2.py & C/train_catboost_v2.py`

```
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score

y_true = pd.read_csv("../data/train.csv", usecols=["is_fraud"])["is_fraud"].values
for exp_id in ["glm_c_v1", "cat_c_v2"]:
    oof = np.load(f"outputs/{exp_id}_oof.npy")
    print(f"{exp_id} Overall OOF ROC-AUC: {roc_auc_score(y_true, oof):.5f}")
```
